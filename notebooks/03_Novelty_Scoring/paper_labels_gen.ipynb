```python
import os
import json
import pandas as pd

labels = []

mapping_base = "../mapped_json/"

for file in os.listdir(mapping_base):
    if not file.endswith(".json"):
        continue

    # Determine label from filename
    if "_novel" in file:
        label = 1
    else:
        label = 0

    # Extract topic (e.g., dia, mt, qa, etc.)
    parts = file.replace(".json", "").split("_")
    topic = parts[1].upper()  # dia → DIA

    with open(os.path.join(mapping_base, file), "r") as f:
        mapping = json.load(f)

    for pid in mapping.keys():
        # Construct global ID
        if label == 1:
            global_id = f"NOVEL_{topic}_{pid}"
        elif "_blog" in file:
            global_id = f"BLOG_{topic}_{pid}"
        else:
            global_id = f"SKG_{topic}_{pid}"

        labels.append((global_id, label))

label_df = pd.DataFrame(labels, columns=["paper_id", "label"])
label_df = label_df.drop_duplicates()

label_df.to_csv("../outputs/paper_labels.csv", index=False)

print(label_df["label"].value_counts())
```
